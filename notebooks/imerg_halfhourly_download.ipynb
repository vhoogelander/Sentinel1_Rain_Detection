# IMERG half-hourly download for a spatial subset
Downloads GPM_3IMERGHH (V07) from NASA GES DISC one day at a time, crops to a bounding box, saves one compressed NetCDF per day, and deletes the raw global files.

**Before running:** create an Earthdata account and approve the *NASA GESDISC DATA ARCHIVE* app in your profile.

```python
import shutil
from datetime import datetime, timedelta
from pathlib import Path

import earthaccess
import xarray as xr
from Sentinel_Rain_Detection.data_downloads import download_imerg_subset
```

## Authenticate

```python
auth = earthaccess.login(strategy="all")
print("Authenticated:", auth.authenticated)
```

## Settings

```python
MinLon, MinLat, Maxlon, MaxLat = -2.70, 10.27, -2.40, 10.57 #(west, south, east, north)
start_time = "2026-01-01"
end_time = "2026-01-02"
imerg_version = 'late'
```

```python
# bbox = (west, south, east, north)
files = download_imerg_subset(
    start=start_time, 
    end=end_time,
    bbox=(MinLon, MinLat, Maxlon, MaxLat),  ##MinLon, MinLat, Maxlon, MaxLat
    run=imerg_version, ## IMERG version (final, late or early)
    variables=("precipitation",),
    outdir="data/imerg_out",
    merge=True,
)
```
